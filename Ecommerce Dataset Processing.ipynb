# Processed E-commerce Dataset

Finally, the processed dataset will be exported as a CSV file.

##### Import Required Library

```python
import pandas as pd
print("Pandas imported successfully!")
```

##### Load the Datasets
*The e-commerce data is provided in three separate CSV files: Orders, Customers, and Products.*

```python
orders = pd.read_csv("Orders.csv")
customers = pd.read_csv("Customers.csv")
products = pd.read_csv("Products.csv")
```

```python
# Preview the datasets
print("Orders Dataset:")
display(orders.head())

print("Customers Dataset:")
display(customers.head())

print("Products Dataset:")
display(products.head())
```

##### Explore the Datasets

```python
# Check the shape of each dataset
print("Orders Shape:", orders.shape)
print("Customers Shape:", customers.shape)
print("Products Shape:", products.shape)

# Check the column names
print("\nOrders Columns:", orders.columns.tolist())
print("Customers Columns:", customers.columns.tolist())
print("Products Columns:", products.columns.tolist())
```

```python
# Display dataset information
print("Orders Dataset Information:")
orders.info()

print("\nCustomers Dataset Information:")
customers.info()

print("\nProducts Dataset Information:")
products.info()
```

```python
# Checking for Missing Values

print("Missing values in Orders:")
print(orders.isnull().sum())

print("\nMissing values in Customers:")
print(customers.isnull().sum())

print("\nMissing values in Products:")
print(products.isnull().sum())
```

##### Merge Orders and Customers

*The Orders and Customers datasets are related through the `Customer_ID` column. We will use a left join to combine the customer information with the order records.*

```python
# Merge Orders with Customers using Customer_ID
orders_customers = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)

# Preview the merged dataset
print("Shape:", orders_customers.shape)
print("Columns:", orders_customers.columns.tolist())

orders_customers.head()
```

##### Merge with Products

*The merged dataset is now combined with the Products dataset using `Product_ID` to add product-related information.*

```python
# Merge the dataset with Products using Product_ID
processed_data = pd.merge(
    orders_customers,
    products,
    on="Product_ID",
    how="left"
)

# Preview the processed dataset
print("Shape:", processed_data.shape)
print("Columns:", processed_data.columns.tolist())

processed_data.head()
```

##### Concatenate DataFrames

*The `concat()` function is used to combine DataFrames along a particular axis.*

*Here, the Orders dataset is divided into two parts and then concatenated vertically to demonstrate the use of `concat()`.*

```python
# Split the Orders DataFrame into two parts
orders_part1 = orders.iloc[:60]
orders_part2 = orders.iloc[60:]

# Concatenate the two parts
orders_combined = pd.concat(
    [orders_part1, orders_part2],
    axis=0,
    ignore_index=True
)

# Preview the concatenated DataFrame
orders_combined.head()
```

```python
# Check the shape after concatenation
orders_combined.shape
```

##### Create a Derived Column using `apply()`

*The `Quantity` and `Unit_Price` columns are used to calculate the total value of each order.*

*The `apply()` function is used to create a new `Order_Value` column.*

```python
# Calculate the total value of each order
processed_data["Order_Value"] = processed_data.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)

# Preview the updated dataset
processed_data.head()
```

##### DateTime Operations

*The `Order_Date` column is converted into DateTime format to extract useful information such as month, day, and day of the week.*

```python
# Convert Order_Date to DateTime format
processed_data["Order_Date"] = pd.to_datetime(processed_data["Order_Date"])

# Extract date-related information
processed_data["Order_Month"] = processed_data["Order_Date"].dt.month
processed_data["Order_Day"] = processed_data["Order_Date"].dt.day
processed_data["Order_DayOfWeek"] = processed_data["Order_Date"].dt.day_name()

processed_data.head()
```

##### Demonstrate `concat()`

*The `concat()` function is used to combine DataFrames with the same structure.
A small sample of the processed dataset is split into two parts and then concatenated back together.*

```python
# Split the processed dataset into two parts
part1 = processed_data.iloc[:5]
part2 = processed_data.iloc[5:10]

# Combine the two parts using concat()
concat_demo = pd.concat([part1, part2], ignore_index=True)

# Display the concatenated DataFrame
concat_demo
```

##### Organize the Final Processed Dataset

```python
# Arrange the columns in an order
processed_data = processed_data[
    [
        "Order_ID",
        "Order_Date",
        "Order_Month",
        "Order_Day",
        "Order_DayOfWeek",
        "Customer_ID",
        "Customer_Name",
        "City",
        "Region",
        "Membership_Type",
        "Product_ID",
        "Product_Name",
        "Category",
        "Brand",
        "Quantity",
        "Unit_Price",
        "Order_Value",
        "Payment_Method",
        "Order_Status"
    ]
]

# Verify the final processed dataset
print("Shape:", processed_data.shape)
print("Columns:", processed_data.columns.tolist())

processed_data.head()
```

##### Export CSV

```python
# Export the processed dataset as a CSV file
processed_data.to_csv("Processed_Ecommerce_Dataset.csv", index=False)

print("Processed dataset exported successfully.")
```
